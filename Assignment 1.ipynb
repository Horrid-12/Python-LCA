# Lists

```python
fruits = ["Orange", "Apple", "Mango", "Banana"]
print (fruits)

# Access List

print(fruits[1])
print("Fruit number 2 is", fruits[1])
# Change list Items

fruits[3] = "Pineapple"

print("Changed Fruit is", fruits[3])
# Added Butterscotch
fruits.insert(4, "Butterscotch")
print("Added" ,fruits[4])

fruits.remove("Apple")
print("Removed Apple")

# Sorting

fruits.reverse()
print("Reversing the order" ,fruits)

# Copying List
fruitscopy = fruits.copy()
print("Copied" ,fruits)
# Joining List
Sweets = ["Candy", "Chocolate"]

Food = fruits + Sweets
print("Food Items are:" ,Food)


fruits.extend(Sweets)
print(fruits)

print("Food Items are:" ,Food)


for x in fruits:
    print ("List of Food Items:",x)

#Clear Lists

fruits.clear()
print(fruits)

```

# Tuple
