```python
#  Table of Two
i = 2
while i <= 20:
    print(i)
    i += 2
    print(i)
    i += 2
```

```python
# Displaying numbers 1 to 10
i = 1
while i <= 10:
    print(i)
    i += 1
```

```python
# Diplaying numbers from 10 to 1
i = 10
while i >= 1:
    print(i)
    i -= 1
```

```python
# Table of a user input number
num = int(input("Enter a number to display its multiplication table: "))
i = 1
while i <= 10:  
    print(f"{num} x {i} = {num * i}")
    i += 1
```

```python
# Squares of user input number
num = int(input("Enter a number to display its square: "))
i = 1
while i <= 1:
    print(f"{num} x {num} = {num * num}")
    i += 1
```

```python
# Exponent of a user input number
base = int(input("Enter the base number: "))
exponent = int(input("Enter the exponent: "))
result = 1
print(f"{base} raised to the power of {exponent} is: ", end="")
while exponent > 0:
    result *= base
    exponent -= 1
print(result)
```
