```python
# Initial List
students = ["Aarav", "Isha", "Rohan", "Sneha", "Kabir"]

# 3rd Student in the list
third_student = students[2]
print("The 3rd student in the list is:", third_student)
# Correcting Spelling Error of Rohan
students[2] = "Rohan Patil"
print("Updated List of Students:", students)
# Adding two new students Ananya and Vihaan
students.append("Ananya")
students.append("Vihaan")
print("List of Students after adding new students:", students)
#Removing Kabir from the list
students.remove("Kabir")
print("List of Students after removing Kabir:", students)
#Total number of Registered Students
total_students = len(students)
print("Total number of registered students:", total_students)   
# Checking if Isha is in the list
if "Isha" in students:
    print("Isha is registered in the list of students.")    
else:   
    print("Isha is not registered in the list of students.")
#Arranging list in alphabetical order
students.sort()
print("List of Students in alphabetical order:", students)
#Backing up the list of students
backup_students = students.copy()
print("Backup of the list of students:", backup_students)
# Adding List of Students from AI Hackathon

hackathon_students = ["Meera", "Arjun", "Isha"]
print("List of Students from AI Hackathon:", hackathon_students)
# Merging the two lists
students.extend(hackathon_students)
print("Merged List of Students:", students)
#Displaying all the lists individually
print("Final List of Students:", students)
print("Backup List of Students:", backup_students)
print("List of Students from AI Hackathon:", hackathon_students)
```
