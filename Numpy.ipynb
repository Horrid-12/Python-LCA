```python
import numpy as np

arr = np.array([[[1, 2, 6], [78, 12, 0]]])
arr2 = np.array([[[1, 34, 1], [13, 16, 24]]])

sum = arr + arr2

print("Sum of Arrays are:", sum)
print("2nd column in Array 1 is:", arr[0, 1])
print(arr[0, 0])
# Copying Sum
x = sum.copy()
print(x)

# Shapes of Arrays
print("Shape of Sum:", sum.shape)
# Creating Arrays Using ndmin
arrn = np.array([17, 34, 132], ndmin = 5)
print(arrn)

arrnr = arrn.reshape(3)
print(arrnr)
sarr = np.array_split(arr, 3)
print(sarr)
```
