```python
# Initial Courses
Courses = ("Python", "Data Science", "Cloud Computing", "Cyber Security", "AI")
# Second Course Selected by the student
print("Second Course Selected by the student is:", Courses[1])
# Last Two Courses selected by the students
print("Last Two Courses Selected by the students are:",Courses[3:])
#Check Weather Python is Present in Courses
is_Python_present = "Python" in Courses
print("Is 'Python' present?", is_Python_present)
# Finding Position of Cloud Computing
cloud_index = Courses.index("Cloud Computing")
print("Position of Cloud Computing:", cloud_index)
#Total Number of Courses
total_courses = len(Courses)
print("Total Number of Courses:", total_courses)
# Adding Machine Learning and Web Development
Courses = Courses + ("Machine Learning", "Web Development")
print("After adding new courses:", Courses)
#Drop Cyber Security
cyber_index = Courses.index("Cyber Security")
Courses = Courses[:cyber_index] + Courses[cyber_index+1:]
print("After Dropping Cyber Security:", Courses)
#Arranging Courses 
sort_courses = tuple(sorted(Courses))
print("After Sorting Courses", sort_courses)
#Backup of Courses
backup_courses = sort_courses
#Combining Student's Courses and Friend's Courses
Friend_Course = ("AI", "Blockchain", "Python")
Combined_Courses = Courses + Friend_Course
#Displaying Final Tuples
print("Final Course Tuple:", Courses)
print("Sorted Tuple:", sort_courses)
print("Backup Tuple:", backup_courses)
print("Combined_Tuple", Combined_Courses)
```
